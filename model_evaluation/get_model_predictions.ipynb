```python
import pandas as pd
import joblib
```

```python
data = pd.read_csv("path/to/data.csv")
model = joblib.load("path/to/model.joblib")
scaler = joblib.load("path/to/scaler.joblib")
```

```python
cols_to_scale = list(scaler.feature_names_in_)

# Scale only the selected columns
scaled_data = pd.DataFrame(
    scaler.transform(data[cols_to_scale]),
    columns=cols_to_scale,
    index=data.index
)

# Keep the remaining columns unchanged
unscaled_df = data.drop(columns=cols_to_scale)

# Concatenate them back together
scaled_data = pd.concat([scaled_data, unscaled_df], axis=1)
scaled_data = scaled_data[model.feature_names_in_]
```

```python
predictions = model.predict_proba(scaled_data)[:,1]
```
